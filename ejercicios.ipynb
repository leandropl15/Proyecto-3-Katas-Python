### Ejercicio 1

- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def frecuencia_letras(cadena):
    frecuencias = {}

    for caracter in cadena:
        if caracter != " ":
            if caracter in frecuencias:
                frecuencias[caracter] += 1
            else:
                frecuencias[caracter] = 1

    return frecuencias

texto = "hola mundo"
print(frecuencia_letras(texto))
```

### Ejercicio 5

- Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
#Defino el parametro opcional nota_aprobado, calculo la media usando sum() y len() y comparo el resultado con la nota de aprovacion

def evaluar_notas(lista, nota_aprobado=5):
    media = sum(lista) / len(lista)

    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"

    return media, estado


notas = [6, 8, 5, 7, 4]

resultado = evaluar_notas(notas)
print(resultado)
```

```python
def factorial(numero):
    if numero == 0 or numero == 1:
        return 1
    else:
        return numero * factorial(numero - 1)



print(factorial(5))
```

- Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.


```python
#Se crea una excepción personalizada para identificar  el caso en el que la lista está vacía. Si ocurre este caso, se lanza con  raise y se captura con try-except para manejar el error.

class ListaVaciaError(Exception):
    """Excepción personalizada para listas vacías."""
    pass


def calcular_promedio(lista):
    if len(lista) == 0:
        raise ListaVaciaError("La lista está vacía.")

    return sum(lista) / len(lista)


try:
    numeros = [4, 6, 8, 10]

    promedio = calcular_promedio(numeros)
    print(f"El promedio es: {promedio}")

except ListaVaciaError as error:
    print(error)
```

### Ejercicio 17

- Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().

```python
# utlizo reduce() para combinar los digitos en un numero unico, el acumulador se multiplica por 10 en cada iteracion y por ultimo se suma el siguiente numero asi se construye de izquierda a derecha

#Algorimo: utilizo el acumulador en cada iteracion al multiplicar por 10 las cifras se desplazan a la izquierda y deja libra la posicion de las unidades esto permite que al sumar el siguiente donde esta el 0 quede en la posicion y asi se va construyendo el numero de izquierda a derecha.

from functools import reduce

def lista_a_numero(lista):
    return reduce(lambda acumulador, digito: acumulador * 10 + digito, lista)


digitos = [4, 5, 3, 6]

print(lista_a_numero(digitos))
```

### Ejercicio 28

- Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.



```python
def primer_duplicado(lista):
    vistos = set()

    for elemento in lista:
        if elemento in vistos:
            return elemento
        vistos.add(elemento)

    return None


numeros = [3, 7, 2, 5, 7, 8, 2]

print(primer_duplicado(numeros))
```

### Ejercicio 34

- Solucion: creo la clase arbol con dos atributos tronco y ramas. Ambos modifican el estado del objeto segun no pedido en el enunciado, y info_arbol() devuelve un resumen con la informacion del arbol

```python
class Arbol:
    def __init__(self):
        self.tronco = 1
        self.ramas = []

    def crecer_tronco(self):
        self.tronco += 1

    def nueva_rama(self):
        self.ramas.append(1)

    def crecer_ramas(self):
        for i in range(len(self.ramas)):
            self.ramas[i] += 1

    def quitar_rama(self, posicion):
        if 1 <= posicion < len(self.ramas):
            self.ramas.pop(posicion -1)

    def info_arbol(self):
        return {
            "Longitud del tronco": self.tronco,
            "Número de ramas": len(self.ramas),
            "Longitudes de las ramas": self.ramas
        }



arbol = Arbol()

arbol.crecer_tronco()
arbol.nueva_rama()
arbol.crecer_ramas()
arbol.nueva_rama()
arbol.nueva_rama()
arbol.quitar_rama(1)

print(arbol.info_arbol())
```

### Ejercicio 35

- Ojo mensaje de error bob tiene saldo insuficiente como indica lanzar el error en caso de fallo.

```python
class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta_corriente):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("Saldo insuficiente.")
        self.saldo -= cantidad

    def agregar_dinero(self, cantidad):
        self.saldo += cantidad

    def transferir_dinero(self, otro_usuario, cantidad):
        if cantidad > otro_usuario.saldo:
            raise ValueError(f"{otro_usuario.nombre} no tiene saldo suficiente.")

        otro_usuario.retirar_dinero(cantidad)
        self.agregar_dinero(cantidad)


# Caso de uso
alicia = UsuarioBanco("Alicia", 100, True)
bob = UsuarioBanco("Bob", 50, True)

bob.agregar_dinero(20)

try:
    alicia.transferir_dinero(bob, 80)
except ValueError as e:
    print(e)

alicia.retirar_dinero(50)

print(alicia.saldo)
```

### Ejercicio 36


```python
def contar_palabras(texto):
    palabras = texto.split()
    contador = {}

    for palabra in palabras:
        contador[palabra] = contador.get(palabra, 0) + 1

    return contador


def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    return texto.replace(palabra_original, palabra_nueva)


def eliminar_palabra(texto, palabra):
    palabras = texto.split()
    palabras = [p for p in palabras if p != palabra]
    return " ".join(palabras)


def procesar_texto(texto, opcion, *args):
    if opcion == "contar":
        return contar_palabras(texto)

    elif opcion == "reemplazar":
        return reemplazar_palabras(texto, args[0], args[1])

    elif opcion == "eliminar":
        return eliminar_palabra(texto, args[0])

    else:
        raise ValueError("Opción no válida.")



texto = "este ejercicio si me tomo mucho mas tiempo"

print(procesar_texto(texto, "contar"))

print(procesar_texto(texto, "reemplazar", "ejercicio", "kata"))

print(procesar_texto(texto, "eliminar", "mas"))
```

### Ejercicio 39

- Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

```python
import math

def calcular_area(figura, datos):
    if figura == "rectangulo":
        base, altura = datos
        return base * altura

    elif figura == "circulo":
        radio = datos[0]
        return math.pi * radio ** 2

    elif figura == "triangulo":
        base, altura = datos
        return (base * altura) / 2

    else:
        raise ValueError("Figura no válida.")



print(calcular_area("rectangulo", (5, 4)))
print(calcular_area("circulo", (3,)))
print(calcular_area("triangulo", (6, 2)))
```
